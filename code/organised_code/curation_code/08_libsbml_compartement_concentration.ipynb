## Imports

```python
from libsbml import *
import os
```

## Paths

```python
model_dir = '/home/emma/Dokumente/thesis/Model_generation_curation/Curated_models/mb1_mdr_rdr_dp_mb2'

save_dir = '/home/emma/Dokumente/thesis/Model_generation_curation/Curated_models/mb1_mdr_rdr_dp_mb2_lib/'
```

## libsbml configurations

```python
reader = SBMLReader()
writer = SBMLWriter()
```

## Variables

```python
compartment_ids_dict = {
    "C_c": "c",
    "C_p": "p",
    "C_e": "e",
}

old_compartment_ids_list = [
    "C_c",
    "C_p",
    "C_e",
]

default_concentration = 0.0
```

## Main

```python
for file in os.listdir(model_dir):

    print(file)

    document = reader.readSBML(os.path.join(model_dir, file))

    model = document.getModel()

    # Check if the model is valid
    if document.getNumErrors() > 0:
        print(f"Errors in model {file}:")
        for i in range(document.getNumErrors()):
            print(document.getError(i).getMessage())
        continue
    
    for old_id in old_compartment_ids_list:

        compartment = model.getCompartment(old_id)

        # Check if the compartment exists
        if compartment is not None:
            
            # Get the new ID
            new_id = compartment_ids_dict[old_id]

            # Set the new ID
            success = compartment.setId(new_id)
        
            if success == LIBSBML_OPERATION_SUCCESS:
                
                # Update references to this compartment in species, etc.
                for i in range(model.getNumSpecies()):
                    species = model.getSpecies(i)
                    if species.getCompartment() == old_id:
                        species.setCompartment(new_id)

    # Set the initial concentration for all metabolites
    # This could be done in the for loop before but I think it's better 
    # to make this independent from any if condition
    for i in range(model.getNumSpecies()):
        species = model.getSpecies(i)

        species.setInitialConcentration(default_concentration)

    writer.writeSBMLToFile(document, save_dir+file[:-4]+'_lib.xml')
```
