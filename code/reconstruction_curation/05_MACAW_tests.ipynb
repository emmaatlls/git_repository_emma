## Comment
This script only runs the diphophate and duplicate test of the MACAW test suite.  
The output are csv files saved in save_dir. It will save these files as "model_id_phase_1.csv".  
So if your model filename was "my_model.xml" the results will be saved as "my_model_phase_1.csv"

## Imports

```python
import cobra
from cobra.io import read_sbml_model
from macaw.main import *
import os
```

## Paths

```python
# Path to your models
model_dir = '/home/emma/Dokumente/thesis/Model_generation_curation/Curated_models/mb1_mdr/'

# Directory in which you want to save the resulting csv files
save_dir = '/home/emma/Dokumente/thesis/Model_generation_curation/Curated_models/MACAW_results'
```

## Redox_pairs, proton_ids, phosphate_ids

```python
redox_pairs_all = [
    # NAD(H), NADP(H), FAD(H2), and FMN(H2)
    ('nad_c', 'nadh_c'), ('nadp_c', 'nadph_c'), ('fad_c', 'fadh2_c'),
    ('fmn_c', 'fmnh2_c'),
    # ubiquinone-8 and ubiquinol-8
    ('q8_c', 'q8h2_c'),
    # menaquinone-8 and menaquinol-8
    ('mqn8_c', 'mql8_c'),
    # riboflavin
    ('ribflv_c', 'rbflvrd_c'),
    # glutathione
    ('gthox_c', 'gthrd_c'),
    # glutaredoxin
    ('grxox_c', 'grxrd_c'),
    # thioredoxin
    ('trdox_c', 'trdrd_c'),
    # flavodoxin (reduced and semi-oxidized)
    ('flxso_c', 'flxr_c'),
    # oxygen and hydrogen peroxide
    ('o2_c', 'h2o2_c')
    # 
]
```

```python
# protons in all compartments
proton_ids = ['h_c', 'h_e', 'h_p']
```

```python
# Phosphate in all compartments
pi_ids=["pi_c", "pi_p", "pi_e"]
ppi_ids=["ppi_c", "ppi_p", "ppi_e"]
```

## Main

```python
for file in os.listdir(model_dir):
    if file.endswith('_mb1_mdr.xml'):
        redox2test = []
        model = read_sbml_model(os.path.join(model_dir, file))
        (diphosphate_results) = diphosphate_test(model, ppi_ids, pi_ids)
        for pair in redox_pairs_all: #only test with pairs that are included in the model
            if pair[0] in model.metabolites and pair[1] in model.metabolites:
                    redox2test.append(pair)
                    #print(redox2test)
        (duplicate_results, duplicate_edge_list) = duplicate_test(model, redox2test, proton_ids)

        phase_1_results = duplicate_results.merge(diphosphate_results)

        phase_1_results.to_csv(os.path.join(save_dir, file.replace('.xml', '_macaw.csv')), index=False)
        
        print(f"Finished processing {file} testing redox pairs: {redox2test}")
```
