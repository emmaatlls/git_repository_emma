## Imports

```python
import cobra
from cobra.io import read_sbml_model, write_sbml_model
import os
```

## Paths

```python
model_dir = '/home/emma/Dokumente/thesis/Model_generation_curation/Curated_models/mb1_mdr_rdr_dp_mb2_lib/'

save_dir = '/home/emma/Dokumente/thesis/Model_generation_curation/Curated_models/mb1_mdr_rdr_dp_mb2_lib_bz/'
```

## Items to remove

```python
metabolites = ['bz_p']

reactions = ['BZtex', 'BZt1pp', 'BZt2_pp']
```

## Main

```python
for file in os.listdir(model_dir):

    if f'{file[:-4]}_bz.xml' not in os.listdir(save_dir):
        if not file.endswith(('.xml', '.sbml')):
            continue
        
        model = read_sbml_model(model_dir+file)

        print(model.id)

        if 'bz_p' in model.metabolites:
            model.remove_metabolites([model.metabolites.get_by_id('bz_p')])
            print('bz_p removed')

        for rxn in reactions:
            if rxn in model.reactions:
                model.remove_reactions([model.reactions.get_by_id(rxn)])
                print(f'{rxn} removed')

        write_sbml_model(model, save_dir+f'{file[:-4]}_bz.xml')
```
